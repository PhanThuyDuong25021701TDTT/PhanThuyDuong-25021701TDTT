```python
# bài 4:nhập 1 dòng gồm các cặp key: value cách nhau bởi dấu cách. Tạo dic ánh xạ mối key thành list các value theo thứ tự
s = input()
a = s.split()               # các cặp giá trị cách nhau bởi 1 dấu cách
d = {}
for x in a:
    if ':' in x:
        key, val= x.split(':',1)
        if key not in d:
            d[key]= []
        else:
            d[key].append(val)
print(d)
```

```python
#bài 5: nhập 1 list số nguyên, in ra dic, pos là dương, nag là âm, =0 thì zeros
s = list(map(int, input().split()))
dic= { 'pos':[], 'nag':[], 'zeros':[] }
for x in s:
    if x >0:
        dic['pos'].append(x)
    elif x<0:
        dic['nag'].append(x)
    elif x ==0:
        dic['zeros'].append(x)
print(dic)
```

```python
# bài 8:nhập 1 list các chuỗi, đếm số lần xuất hiện của từng chuỗi và lưu vào dic
s = input().split()
dic={}
for x in s:
    if x in dic:
        dic[x] +=1
    else:
        dic[x] =1      # nếu x không in dic thì lần đầu tiên sẽ +1 trước
print(dic)
    
```

```python
#bài 9: nhập 2 dic, gộp 2 dic, nếu có 1 gtri thì giữ nguyên, có xuất hiện trong 2 dic thì cộng lại
dic1= {}
dic2={}
d ={}
n = int(input())
m = int(input()) 
for i in range(n):
    k,v = map(int, input().split())
    dic1[k]=v
for i in range(m):
    k,v =map( int, input().split())
    dic2[k]=v
for k in dic1:
    d[k]= dic1[k]
for k in dic2:
    if k in dic1:
        d[k] += dic2[k]
    else:
        d[k]= dic2[k]
dic= sorted(d)
for k in dic:
    print( k, d[k])
    


```

```python
# bài 10:nhập 1 list số nguyên và 1 số nguyên dương k. lấy  2 trong list số có tổng =k rồi cho vào 1 tuple và xếp theo thứ tự tăng dần của số đầu tiên trong tuple
s= list(map(int,input().split()))
k=int(input())
a=[]
for i in range(len(s)):
    for j in range( i+1, len(s)):
        if s[i]+s[j]==k:
            a.append((s[i],s[j]))
def lay_phan_tu_dau(x):
    return x[0]
a.sort(key=lay_phan_tu_dau)  # không thêm (x) vì nó là gọi hàm thôi rồi python tự chạy, còn cho x là x chx định nghĩa nhưng đã bắt nó chạy là ko đc
print(a)
            
```

```python
# bài 12:nhập vào 1 list số nguyên. tìm số có xuất hiện nhiều nhất. cùng số lần xuất hiện thì in số nhỏ hơn
s = list(map(int, input().split()))
d=0
a=0

for i in range(len(s)):
    cnt=0
    for j in range(len(s)):
        if s[i] == s[j]:
            cnt+=1
    if cnt>d:
        d=cnt
        a= s[i]
    elif d ==cnt: 
        d==cnt
        a= min(a, s[i])
print(a)
    




```

```python
#bài 14: 2 dòng, 2 danh sách. In ra danh sách mới chứa các phần tử chung, ko trùng lặp
a= list(map(int, input().split()))
b=list(map(int, input().split()))
s=[]
for i in a:
    s.append(i)
for j in b:
    if j not in s:
        s.append(j)
s = list(set(s))
print(s)
```
