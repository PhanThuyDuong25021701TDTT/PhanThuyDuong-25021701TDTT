```python
#bài 1:viết hàm tìm kiếm nhị phân trên một danh sách đã dược sắp xếp tăng dần. 
def binary_search(a, x):
    left=0
    right= len(a)-1
    while right>= left:
        mid = (left+right)//2
        if a[mid]==x:
            return mid
        elif a[mid] > x:
            right= right -1
        elif a[mid]<x:
            left= left +1
    return -1
a = list(map(int, input().split()))
x= int(input())
print(binary_search(a, x))
```

```python
def count_occurrences(arr, x):
    dem = 0
    for i in arr:
        if i == x:
            dem += 1
    return dem
arr= list(map(int, input().split()))
x= int(input())
print(count_occurrences(arr, x))
```

```python
#bài 3: bubble sort
s= list(map(int, input().split()))
for i in range(len(s)-1):
    for j in range(len(s)-1):
        if s[j]> s[j+1]:
            s[j], s[j+1]= s[j+1], s[j]
print(s)
```

```python
#bài 3: s= list(map(int, input().split()))
for i in range(len(s)-1):
    min_idx=i
    for j in range(i+1, len(s)):
        if s[j]<s[min_idx]:
            min_idx=j
    s[i], s[min_idx]=s[min_idx], s[i]
print(s)
```
